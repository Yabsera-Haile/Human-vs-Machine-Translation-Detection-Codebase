# Human vs Machine Translation Detection Codebase

This Jupyter Notebook is part of the "Human-vs-Machine-Translation-Detection-Codebase" repository, aimed at analyzing and comparing translation quality metrics to distinguish between human and machine-generated translations.

## Overview
The notebook includes functions to load translation data from CSV files and compute statistical summaries of key metrics such as BLEU, BLEURT, and COMET scores. It provides overall statistics and breakdowns grouped by source language, facilitating insights into translation performance across different languages and systems.

## Key Components
- **Data Loading**: Reads CSV files containing translation metrics.
- **Metric Analysis**: Identifies and aggregates scores for BLEU, BLEURT, and COMET.
- **Statistical Summaries**: Calculates mean and standard deviation for overall data and per source language group.

This codebase supports research in natural language processing, particularly in evaluating translation quality and detecting machine-generated text.

```python
import pandas as pd

def analyze_translation_metrics(csv_path):
    # Load data
    df = pd.read_csv(csv_path)

    # Identify metric columns
    bleu_cols = [col for col in df.columns if col.endswith("_bleu")]
    bleurt_cols = [col for col in df.columns if col.endswith("_bleurt")]
    comet_cols = [col for col in df.columns if col.endswith("_comet")]

    print("=== OVERALL STATISTICS ===\n")

    def print_stats(name, cols):
        print(f"{name}:")
        stats = df[cols].agg(['mean', 'std'])
        print(stats)
        print()

    print_stats("BLEU", bleu_cols)
    print_stats("BLEURT", bleurt_cols)
    print_stats("COMET", comet_cols)

    # Grouped by source language
    print("\n=== STATISTICS BY SOURCE LANGUAGE ===\n")

    def print_grouped_stats(name, cols):
        print(f"{name} (grouped by source language):")
        grouped = df.groupby("source language")[cols].agg(['mean', 'std'])
        print(grouped)
        print()

    print_grouped_stats("BLEU", bleu_cols)
    print_grouped_stats("BLEURT", bleurt_cols)
    print_grouped_stats("COMET", comet_cols)
```
